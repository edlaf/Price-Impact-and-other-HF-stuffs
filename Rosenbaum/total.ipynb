```python
data.price_impact()
price_impact = data.d_price
queue_size = data.d_size
ws = 100
price_impact = [np.mean(price_impact[i:i+ws]) for i in range (len(price_impact-ws))]
print(np.mean(price_impact))
queue_size = [np.mean(queue_size[i:i+ws]) for i in range (len(queue_size-ws))]
fig = go.Figure()
fig.add_trace(go.Scatter(x=queue_size, y=queue_size+3*np.ones(len(queue_size)), mode='lines', name="id", line=dict(width = 0.85, color = 'darkblue')))
fig.add_trace(go.Scatter(x=queue_size, y = price_impact, mode='lines', name="PI", line=dict(width = 0.85, color = 'darkblue')))
fig.add_trace(go.Scatter(x=queue_size, y=np.array(queue_size)**(1/2)+2*np.ones(len(queue_size)), mode='lines', name="id", line=dict(width = 0.85, color = 'darkblue')))

fig.update_layout(
    title="Number of events N+ and N-",
    xaxis=dict(
        type="log",
        showgrid=True,
        gridcolor='#808080'
    ),
    yaxis=dict(
        type="log",
        showgrid=True,
        gridcolor='#808080'
    ),
    plot_bgcolor='#D3D3D3',
    paper_bgcolor='#D3D3D3'
)
fig.show()
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=queue_size[28800:29050], y = 27*np.ones(len(queue_size[28800:29050]))*queue_size[28800:29050], mode='lines', name="identity", line=dict(width = 2, color = 'red')))
fig.add_trace(go.Scatter(x=queue_size, y = price_impact, mode='lines', name="PI", line=dict(width = 0.85, color = 'darkblue')))
fig.add_trace(go.Scatter(x=queue_size[6000:28800], y = 17*np.ones(len(queue_size[6000:28800]))*np.array(queue_size[6000:28800])**(1/2), mode='lines', name = "racine", line=dict(width = 2, color = 'red')))

fig.update_layout(
    width=1800,         # largeur en pixels
    height=1200,  
    xaxis=dict(
        title = "sqrt(Q/V)",
        type="log",
        showgrid=True,
        gridcolor='#808080'
    ),
    yaxis=dict(
        title = "I/sigma",
        type="log",
        showgrid=True,
        gridcolor='#808080'
    ),
    plot_bgcolor='White',
    paper_bgcolor='White'
)
fig.show()
```
